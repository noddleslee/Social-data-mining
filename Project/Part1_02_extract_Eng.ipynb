```python
def read_tsv_from_zip(zip_path, inner_name, columns=None):
    with zipfile.ZipFile(zip_path, "r") as z:
        with z.open(inner_name) as f:
            return pd.read_csv(
                f,
                sep="\t",              # FriendFeed 是 TSV
                names=columns,
                header=None if columns else "infer",
                na_values="\\N",
                engine="python",
                encoding="utf-8",
                on_bad_lines="skip"
            )
```

```python
import re
import html

def clean_text(text):
    if not isinstance(text, str) or text.strip() == "":
        return ""

    # 1) 删除 <a>...</a>（你原写法有个空格：</a >，我这里修正为 </a>）
    text = re.sub(r"<a.*?>.*?</a>", "", text, flags=re.DOTALL | re.IGNORECASE)

    # 2) 干掉其他 HTML 标签
    text = re.sub(r"<.*?>", "", text)

    # 3) 转换 HTML 实体 (如 &amp; -> &)
    text = html.unescape(text)

    # 4) 删除 URL
    text = re.sub(r"(https?://\S+|www\.\S+)", "", text)

    # 5) 删除 RT 转发块（保守写法：避免误删太多）
    text = re.sub(r"RT\s*@.*?:.*?(?= - |$)", "", text, flags=re.IGNORECASE)

    # 6) 删除 via 来源和 @用户名
    text = re.sub(r"via\s*@\S+", "", text, flags=re.IGNORECASE)
    text = re.sub(r"@\w+", "", text)

    # 7) 清理多余空格
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
from langdetect import detect

def is_english(text):
    if not isinstance(text, str) or len(text) < 10:
        return False
    try:
        return detect(text) == "en"
    except:
        return False
```

```python
entires_sub['cleaned_text'] = entires_sub['Text'].apply(clean_text)
entires_sub['cleaned_text'].sample(5).to_list()
```
